# Assignment 1

## Question
Based on what you have learned in class, do the following:
1. Create a new folder.
2. Put Code.txt, Log.txt, and Output.txt in the folder.
3. Stage Code.txt and Output.txt.
4. Commit them.
5. Push them to GitHub.

```
# Create a new folder
mkdir GitAssignment
cd GitAssignment

# Create files
touch Code.txt Log.txt Output.txt

# Stage Code.txt and Output.txt
git add Code.txt Output.txt

# Commit changes
git commit -m "Added Code.txt and Output.txt"

# Add remote repository
git remote add origin <repository-url>

# Push to GitHub
git push -u origin main
```

# Assignment 2

## Question
Create a Git working directory, create branches develop, feature1, feature2, stash develop.txt, work on feature1, restore and commit develop.txt.

```
mkdir Assignment2
cd Assignment2
git init

touch feature1.txt feature2.txt
git add .
git commit -m "Added feature files"

git branch develop
git branch feature1
git branch feature2

git checkout develop
touch develop.txt
git stash

git checkout feature1
touch new.txt
git add new.txt
git commit -m "Added new.txt"

git checkout develop
git stash pop
git add develop.txt
git commit -m "Added develop.txt" 
```

# Assignment 3

## Question
Create branches develop, f1, f2. Commit main.txt, add files in develop, push all branches, and delete f2.

```
mkdir Assignment3
cd Assignment3
git init

git branch develop
git branch f1
git branch f2

touch main.txt
git add main.txt
git commit -m "Added main.txt"

git checkout develop
touch f1.txt f2.txt
git add .
git commit -m "Added f1.txt and f2.txt"

git remote add origin <repository-url>

git push -u origin master
git push origin develop
git push origin f1
git push origin f2

git branch -d f2
git push origin --delete f2
```

# Assignment 4

## Question
Create master.txt, branches public1, public2, private, perform merges and updates.

```
mkdir Assignment4
cd Assignment4
git init

touch master.txt
git add .
git commit -m "Added master.txt"

git branch public1
git branch public2
git branch private

git checkout public1
touch public1.txt
git add .
git commit -m "Added public1.txt"

git checkout master
git merge public1

git checkout public2
touch public2.txt
git add .
git commit -m "Added public2.txt"

git checkout master
git merge public2

git checkout private
echo "Updated content" >> master.txt
git add .
git commit -m "Updated master.txt"

git checkout public1
git merge private

git checkout public2
git merge private

git checkout private
git merge master
```
