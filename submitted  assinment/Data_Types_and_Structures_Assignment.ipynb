Part 1 — Theoretical Questions

### 1. What are data structures, and why are they important?

**Data structures** are specialized formats for organizing, storing, and managing data in a computer so it can be accessed and modified efficiently. They are important because they:
- Allow efficient data access and modification.
- Reduce time and space complexity of algorithms.
- Help in solving real-world problems like searching, sorting, and database management.
- Form the backbone of efficient software design.

### 2. Explain the difference between mutable and immutable data types with examples.

- **Mutable data types** can be changed after creation. Examples: `list`, `dict`, `set`.
- **Immutable data types** cannot be changed after creation. Examples: `int`, `float`, `str`, `tuple`, `bool`.

```python
# Mutable
lst = [1, 2, 3]
lst[0] = 10   # works

# Immutable
s = 'hello'
# s[0] = 'H'  # ❌ TypeError
```

### 3. What are the main differences between lists and tuples in Python?

| Feature | List | Tuple |
|---|---|---|
| Mutability | Mutable | Immutable |
| Syntax | `[1,2,3]` | `(1,2,3)` |
| Performance | Slower | Faster |
| Use case | Data that changes | Fixed data |
| Methods | Many (append, remove…) | Few (count, index) |

### 4. Describe how dictionaries store data.

Dictionaries store data as **key–value pairs** using a **hash table** internally. Each key is hashed to produce an index where its value is stored, enabling **O(1) average-time** lookups, insertions, and deletions. Keys must be immutable (e.g., strings, numbers, tuples).

### 5. Why might you use a set instead of a list in Python?

Use a **set** when:
- You need **unique** elements (no duplicates).
- You need **fast membership tests** (`in` is O(1) on average vs O(n) for lists).
- You need **mathematical set operations** like union, intersection, difference.

### 6. What is a string in Python, and how is it different from a list?

A **string** is an immutable sequence of Unicode characters enclosed in quotes.

Differences from a list:
- Strings are **immutable**; lists are **mutable**.
- Strings hold only characters; lists can hold any data type.
- Strings support text-specific methods (`upper()`, `split()`); lists support `append()`, `remove()`, etc.

### 7. How do tuples ensure data integrity in Python?

Tuples are **immutable**, so once created their elements cannot be modified, added, or removed. This guarantees that the stored data remains unchanged throughout the program — ideal for fixed configurations, coordinates, or database records.

### 8. What is a hash table, and how does it relate to dictionaries in Python?

A **hash table** is a data structure that maps keys to values using a hash function to compute an index into an array of buckets. Python dictionaries are **implemented as hash tables**, which is why they provide average **O(1)** time complexity for lookup, insert, and delete operations.

### 9. Can lists contain different data types in Python?

**Yes.** Python lists are heterogeneous — they can hold elements of any data type, including mixed types in the same list.

```python
mixed = [1, 'hello', 3.14, [1,2], True]
```

### 10. Explain why strings are immutable in Python.

Strings are immutable because:
- It allows **safe sharing** of string objects (memory efficiency via interning).
- It makes them **hashable**, so they can be dictionary keys or set elements.
- It ensures **thread safety** and prevents accidental modification.
- It improves **performance** for string operations.

### 11. What advantages do dictionaries offer over lists for certain tasks?

- **Faster lookups** by key (O(1) vs O(n) for a list search).
- **Meaningful keys** instead of numeric indices.
- Ideal for storing **structured / labeled data** (e.g., records).
- Easier representation of real-world entities (like JSON-like data).

### 12. Describe a scenario where using a tuple would be preferable over a list.

Storing **fixed data** that should not change, such as **geographic coordinates** `(latitude, longitude)`, RGB colors `(255, 0, 0)`, or days of the week. Tuples are also preferable when the data must serve as a **dictionary key** or be stored in a **set**, since tuples are hashable.

### 13. How do sets handle duplicate values in Python?

Sets **automatically remove duplicates** — only unique elements are stored.

```python
s = {1, 2, 2, 3, 3, 3}
print(s)  # {1, 2, 3}
```

### 14. How does the “in” keyword work differently for lists and dictionaries?

- For **lists**: `in` searches through each element sequentially — **O(n)** time.
- For **dictionaries**: `in` checks for the existence of a **key** using the hash table — **O(1)** average time.

### 15. Can you modify the elements of a tuple? Explain why or why not.

**No.** Tuples are immutable, so their elements cannot be modified after creation. Attempting to do so raises a `TypeError`. However, if a tuple contains a **mutable object** (e.g., a list), that inner object can be modified — but the tuple itself still references the same object.

### 16. What is a nested dictionary, and give an example of its use case.

A **nested dictionary** is a dictionary where the values themselves are dictionaries.

**Use case:** storing student records.

```python
students = {
    'S1': {'name': 'Alice', 'age': 20, 'grade': 'A'},
    'S2': {'name': 'Bob',   'age': 22, 'grade': 'B'}
}
```

### 17. Describe the time complexity of accessing elements in a dictionary.

Accessing elements in a dictionary by key has an **average time complexity of O(1)** due to the hash table implementation. In the worst case (many hash collisions), it can degrade to **O(n)**, but this is rare in practice.

### 18. In what situations are lists preferred over dictionaries?

- When the data is an **ordered sequence** without natural keys.
- When you need to **iterate by position/index**.
- When duplicates are allowed and meaningful.
- When using sequence-specific operations like slicing, sorting, or reversing.

### 19. Why are dictionaries considered unordered, and how does that affect data retrieval?

Historically (before Python 3.7), dictionaries did not preserve insertion order — items were accessed by key, not position. From Python **3.7+**, dictionaries **do preserve insertion order**, but they are still conceptually **key-based**, meaning retrieval is done via keys rather than positional indices.

### 20. Explain the difference between a list and a dictionary in terms of data retrieval.

- **List:** Data is retrieved using a **numeric index** (e.g., `lst[2]`) — order matters.
- **Dictionary:** Data is retrieved using a **key** (e.g., `d['name']`) — much faster lookups for labeled data.

Lists are best for sequential access; dictionaries are best for keyed access.
