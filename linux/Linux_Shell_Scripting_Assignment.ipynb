## Task 1: File Management Script

```
#!/bin/bash
mkdir -p backup
cp *.txt backup/ 2>/dev/null
ls > backup/file_list.txt
date >> backup/file_list.txt
echo "Backup completed."
```

## Task 2: System Health Check

```
#!/bin/bash
cpu=$(top -bn1 | grep "Cpu" | awk '{print $2+$4}')
mem=$(free | awk '/Mem/ {printf("%.0f", $3/$2*100)}')
echo "CPU Usage: $cpu%"
echo "Memory Usage: $mem%"
```

## Task 3: User Account Management

```
#!/bin/bash
read -p "Enter username: " user
sudo useradd $user
sudo passwd $user
echo "User $user created."
```

## Task 4: Automated Backup

```
#!/bin/bash
read -p "Directory path: " dir
tar -czf backup_$(date +%F).tar.gz "$dir"
echo "Backup created."
```

## Task 5: Disk Space Monitor

```
#!/bin/bash
usage=$(df / | awk 'NR==2 {print $5}' | tr -d '%')
if [ $usage -gt 80 ]; then
  echo "Warning: Disk usage above 80%"
fi
```

## Task 6: Automated Software Installation

```
#!/bin/bash
packages=(git vim curl)
for pkg in "${packages[@]}"
do
  sudo apt install -y $pkg
done
```

## Task 7: Text File Processing

```
#!/bin/bash
read -p "Enter file: " file
sort "$file" | uniq > cleaned.txt
echo "Duplicates removed."
```
