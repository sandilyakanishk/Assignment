# Task 1: Set Up a CI Pipeline

Objective: Set up a basic CI pipeline using GitHub Actions.

Sample Answer:
Create a GitHub repository, add a Node.js/Python application, and create a workflow file (.github/workflows/ci.yml) that installs dependencies and runs tests automatically whenever code is pushed.

# Task 2: Automated Build with Jenkins

Objective: Configure Jenkins to automatically build a project.

Sample Answer:
Install Jenkins, create a Freestyle Project, connect it to a GitHub repository, configure a webhook trigger, and run build commands such as npm install and npm test.

# Task 3: Unit Testing

Objective: Write unit tests and integrate them into a CI pipeline.

Sample Answer:
Create a small application and write tests using unittest (Python) or Jest (JavaScript). Configure GitHub Actions or Jenkins to execute tests on every push.

# Task 4: Integration Testing

Objective: Test interaction between application components.

Sample Answer:
Create a REST API and write integration tests that verify communication between services, databases, or APIs. Run these tests through the CI pipeline.

# Task 5: Performance Testing

Objective: Automate performance testing in CI/CD.

Sample Answer:
Use k6 or JMeter scripts to simulate user traffic and measure response times. Execute these tests automatically during CI builds.

# Task 6: Security Testing

Objective: Integrate security testing into CI/CD.

Sample Answer:
Use SonarQube for static code analysis and OWASP ZAP for vulnerability scanning. Configure automated scans on every push.

# Task 7: Continuous Deployment

Objective: Automatically deploy to a staging environment.

Sample Answer:
Configure GitHub Actions, Jenkins, or AWS CodeDeploy to deploy the application after successful builds and tests.
